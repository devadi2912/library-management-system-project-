**Tensor data types**

```python
import torch

print (torch.__version__)
if torch.cuda.is_available() :
  print ("gpu is available!")
  print ("using gpu :", torch.cuda.get_device_name(0))
else :
  print ("gpu is not available!")
  print ("using cpu instead")
```

```python
# finding data type of the tensor
x = torch.tensor ([1,4,6], dtype = torch.float32)
y = torch.tensor ([0.115, 2.67, 1.74], dtype = torch.int32)

print (x.dtype)
print (y.dtype) # floor of the float vlaues
```

```python
# assigning datatype while initializing
x = torch.tensor ([1,4,6], dtype = torch.float32)
y = torch.tensor ([0.115, 2.67, 1.74], dtype = torch.int32)

print (x)
print (y)  # displayes the tensor along with the type
```

```python
# changing the data type of an existing tensor
print (x, "is of type", x.dtype)

print (x.to(torch.int32))
```

### PyTorch Tensor Data Types

| Datatype | Dtype | Description |
| :--- | :--- | :--- |
| **32-bit floating point** | `torch.float32` or `torch.float` | Standard single-precision floating-point format for neural networks. |
| **64-bit floating point** | `torch.float64` or `torch.double` | Double-precision floating-point format for precise computations. |
| **16-bit floating point** | `torch.float16` or `torch.half` | Half-precision floating-point format often used for faster GPU training. |
| **16-bit brain floating point** | `torch.bfloat16` | Brain floating-point format optimized for deep learning performance. |
| **8-bit unsigned integer** | `torch.uint8` | Unsigned 8-bit integer, commonly used for image pixel data. |
| **8-bit signed integer** | `torch.int8` | Signed 8-bit integer for low-precision representations. |
| **16-bit signed integer** | `torch.int16` or `torch.short` | Signed 16-bit integer for medium-range whole numbers. |
| **32-bit signed integer** | `torch.int32` or `torch.int` | Standard 32-bit signed integer for indexing and counting. |
| **64-bit signed integer** | `torch.int64` or `torch.long` | Long 64-bit signed integer, standard for target labels and indices. |
| **Boolean** | `torch.bool` | Represents logical True/False values. |
| **64-bit complex** | `torch.complex64` or `torch.cfloat` | Complex numbers represented by two 32-bit float values (real and imaginary). |
| **128-bit complex** | `torch.complex128` or `torch.cdouble` | Complex numbers represented by two 64-bit float values. |
| **Quantized 8-bit signed integer** | `torch.qint8` | Quantized 8-bit signed integer used for post-training quantization. |
| **Quantized 8-bit unsigned integer** | `torch.quint8` | Quantized 8-bit unsigned integer. |
| **Quantized 32-bit signed integer** | `torch.qint32` | Quantized 32-bit signed integer. |
| **Quantized 4-bit signed integer** | `torch.quint4x2` | Packed 4-bit unsigned integers. |
